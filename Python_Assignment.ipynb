# **Sorting and Ranking**

```python
print(
    df.sort_values(
        "predicted_score",
        ascending=False
    )
)
```

```python
print(df.nlargest(10, "predicted_score"))
```

```python
print(df.nsmallest(10, "predicted_score"))
```

```python
print(df.loc[df["predicted_score"].idxmax()])
```

```python
print(df.loc[df["predicted_score"].idxmin()])
```

```python
print(
    df.sort_values(
        "previous_cgpa",
        ascending=False
    )
)
```

```python
print(
    df.sort_values(
        ["semester", "predicted_score"],
        ascending=[True, False]
    )
)
```

```python
df["predicted_score_rank"] = df["predicted_score"].rank(
    ascending=False
)

print(
    df[
        ["student_name", "predicted_score",
         "predicted_score_rank"]
    ]
)
```

```python
print(df.nlargest(10, "engagement_score"))
```

```python
print(
    df.loc[
        df["online_activity_count"].idxmax()
    ]
)
```

# **GroupBy and Aggregation**

```python
print(
    df.groupby("semester")["predicted_score"].mean()
)
```

```python
print(
    df.groupby("semester")["attendance_percent"].mean()
)
```

```python
print(
    df.groupby("gender")["study_hours_per_day"].mean()
)
```

```python
print(df["grade"].value_counts())
```

```python
print(df["risk_level"].value_counts())
```

```python
print(
    df.groupby("python_skill_level")["predicted_score"].mean()
)
```

```python
print(
    df.groupby("mathematics_skill")["previous_cgpa"].mean()
)
```

```python
print(
    df.groupby("semester")["engagement_score"].mean()
)
```

```python
print(
    df.groupby("grade")["predicted_score"]
      .agg(["max", "min"])
)
```

```python
summary = df.groupby("semester").agg(
    average_attendance=("attendance_percent", "mean"),
    average_study_hours=("study_hours_per_day", "mean"),
    average_predicted_score=("predicted_score", "mean"),
    number_of_students=("student_id", "count")
)

print(summary)
```
